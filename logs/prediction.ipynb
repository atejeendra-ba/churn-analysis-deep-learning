```python
import os
import pickle
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import load_model

# Force the exact project root path
BASE_DIR = r'C:\Users\tejee\OneDrive\Documents\ANN Classification Project'

# Load the trained model from project root
model_path = os.path.join(BASE_DIR, 'model.h5')
model = load_model(model_path)

# Load encoders and scaler from project root
with open(os.path.join(BASE_DIR, 'onehot_encoder_geo.pkl'), 'rb') as file:
    onehot_encoder_geo = pickle.load(file)

with open(os.path.join(BASE_DIR, 'label_encoder_gender.pkl'), 'rb') as file:
    label_encoder_gender = pickle.load(file)

with open(os.path.join(BASE_DIR, 'scaler.pkl'), 'rb') as file:
    scaler = pickle.load(file)

print("✅ Model, encoders, and scaler loaded successfully!")
```

```python
import os

project_folder = r'C:\Users\tejee\OneDrive\Documents\ANN Classification Project'
print("Files in project folder:")
print(os.listdir(project_folder))
```

```python
## Example input data for testing
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
## One-hot encode the 'Geography' column
# Use onehot_encoder_geo instead of label_encoder_geo
geo_encoded = onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()

# Create DataFrame with proper feature names
geo_encoded_df = pd.DataFrame(
    geo_encoded, 
    columns=onehot_encoder_geo.get_feature_names_out(['Geography'])
)

# Display the output (single underscore)
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## Encode categorical features
# Encode 'Gender' using label_encoder_gender

input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
# concatination one hot encoded 
input_df = pd.concat([input_df.drop("Geography", axis=1), geo_encoded_df], axis=1)
input_df
```

```python
## scaling the input data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## Predict Churn
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_proba=prediction[0][0]
```

```python
prediction_proba
```

```python
if prediction_proba > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is unlikely to churn.')
```
